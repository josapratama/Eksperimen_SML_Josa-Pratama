# **1. Perkenalan Dataset**


## Dataset: Heart Disease UCI

Dataset ini berasal dari **UCI Machine Learning Repository** dan juga tersedia di Kaggle.

**Deskripsi Dataset:**
Dataset Heart Disease digunakan untuk memprediksi apakah seorang pasien memiliki penyakit jantung berdasarkan berbagai fitur klinis.

**Fitur-fitur Dataset:**
1. `age` - Usia pasien (tahun)
2. `sex` - Jenis kelamin (1 = laki-laki, 0 = perempuan)
3. `cp` - Tipe nyeri dada (0-3)
4. `trestbps` - Tekanan darah saat istirahat (mm Hg)
5. `chol` - Kolesterol serum (mg/dl)
6. `fbs` - Gula darah puasa > 120 mg/dl (1 = ya, 0 = tidak)
7. `restecg` - Hasil elektrokardiografi saat istirahat (0-2)
8. `thalach` - Detak jantung maksimum
9. `exang` - Angina akibat olahraga (1 = ya, 0 = tidak)
10. `oldpeak` - Depresi ST akibat olahraga relatif terhadap istirahat
11. `slope` - Kemiringan segmen ST puncak olahraga (0-2)
12. `ca` - Jumlah pembuluh besar yang diwarnai fluoroskopi (0-3)
13. `thal` - Thalassemia (0-3)
14. `target` - Diagnosis penyakit jantung (1 = penyakit, 0 = normal)

**Sumber:** [Kaggle - Heart Disease UCI](https://www.kaggle.com/datasets/ronitf/heart-disease-uci)

# **5. Data Preprocessing**

Pada tahap ini, kita melakukan data preprocessing untuk mempersiapkan data sebelum digunakan dalam model machine learning.

Tahapan preprocessing yang dilakukan:
1. Menangani Missing Values
2. Menghapus Data Duplikat
3. Deteksi dan Penanganan Outlier (IQR Method)
4. Encoding Data Kategorikal
5. Normalisasi / Standarisasi Fitur Numerik
6. Split Data Train dan Test

```python
# 5.1 Salinan dataset untuk preprocessing
df_processed = df.copy()
print('=== Dataset awal ===')
print(f'Shape: {df_processed.shape}')
```

```python
# 5.2 Menangani Missing Values
print('=== Penanganan Missing Values ===')
print(f'Missing values sebelum: {df_processed.isnull().sum().sum()}')

# Imputasi dengan median untuk kolom numerik jika ada missing values
numerical_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
categorical_cols_feat = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

for col in numerical_cols:
    if df_processed[col].isnull().sum() > 0:
        median_val = df_processed[col].median()
        df_processed[col].fillna(median_val, inplace=True)
        print(f'  Kolom {col}: diisi dengan median = {median_val}')

for col in categorical_cols_feat:
    if df_processed[col].isnull().sum() > 0:
        mode_val = df_processed[col].mode()[0]
        df_processed[col].fillna(mode_val, inplace=True)
        print(f'  Kolom {col}: diisi dengan mode = {mode_val}')

print(f'Missing values sesudah: {df_processed.isnull().sum().sum()}')
```

```python
# 5.3 Menghapus Data Duplikat
print('=== Penghapusan Duplikat ===')
print(f'Shape sebelum: {df_processed.shape}')
df_processed.drop_duplicates(inplace=True)
df_processed.reset_index(drop=True, inplace=True)
print(f'Shape sesudah: {df_processed.shape}')
```

```python
# 5.4 Deteksi dan Penanganan Outlier menggunakan IQR
print('=== Penanganan Outlier (IQR Method) ===')

outlier_cols = ['trestbps', 'chol', 'thalach', 'oldpeak']

def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    before = len(df)
    df = df[(df[column] >= lower) & (df[column] <= upper)]
    after = len(df)
    print(f'  {column}: hapus {before - after} outlier (batas: [{lower:.2f}, {upper:.2f}])')
    return df

shape_before = df_processed.shape
for col in outlier_cols:
    df_processed = remove_outliers_iqr(df_processed, col)

df_processed.reset_index(drop=True, inplace=True)
print(f'\nShape sebelum: {shape_before} -> sesudah: {df_processed.shape}')
```

```python
# 5.5 Encoding Data Kategorikal (One-Hot Encoding)
print('=== Encoding Fitur Kategorikal ===')

# Kolom dengan lebih dari 2 kategori perlu one-hot encoding
multi_cat_cols = ['cp', 'restecg', 'slope', 'ca', 'thal']

df_encoded = pd.get_dummies(df_processed, columns=multi_cat_cols, drop_first=False, dtype=int)

print(f'Shape sebelum encoding: {df_processed.shape}')
print(f'Shape sesudah encoding: {df_encoded.shape}')
print(f'\nKolom setelah encoding:')
print(list(df_encoded.columns))
```

```python
# 5.6 Normalisasi / Standarisasi Fitur Numerik
print('=== Standarisasi Fitur Numerik ===')

# Pisahkan fitur dan target
X = df_encoded.drop('target', axis=1)
y = df_encoded['target']

# Kolom numerik yang akan distandarisasi
num_cols_to_scale = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

scaler = StandardScaler()
X[num_cols_to_scale] = scaler.fit_transform(X[num_cols_to_scale])

print('Standarisasi selesai!')
print(f'\nStatistik fitur numerik sesudah standarisasi:')
print(X[num_cols_to_scale].describe().round(3))
```

```python
# 5.7 Split Data Train dan Test
print('=== Split Data Train dan Test ===')

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_SEED, stratify=y
)

print(f'Total data: {len(X)}')
print(f'Data train: {len(X_train)} ({len(X_train)/len(X)*100:.1f}%)')
print(f'Data test: {len(X_test)} ({len(X_test)/len(X)*100:.1f}%)')
print(f'\nDistribusi target pada train set:')
print(y_train.value_counts())
print(f'\nDistribusi target pada test set:')
print(y_test.value_counts())
```

```python
# 5.8 Simpan dataset hasil preprocessing
import os

output_dir = 'heart_disease_preprocessing'
os.makedirs(output_dir, exist_ok=True)

# Gabungkan X dan y untuk disimpan
df_train = pd.concat([X_train.reset_index(drop=True), y_train.reset_index(drop=True)], axis=1)
df_test = pd.concat([X_test.reset_index(drop=True), y_test.reset_index(drop=True)], axis=1)

df_train.to_csv(f'{output_dir}/train.csv', index=False)
df_test.to_csv(f'{output_dir}/test.csv', index=False)

# Simpan juga dataset preprocessing lengkap
df_full_processed = pd.concat([X.reset_index(drop=True), y.reset_index(drop=True)], axis=1)
df_full_processed.to_csv(f'{output_dir}/heart_disease_preprocessed.csv', index=False)

print('=== Dataset Preprocessing Disimpan ===')
print(f'File train: {output_dir}/train.csv ({df_train.shape})')
print(f'File test: {output_dir}/test.csv ({df_test.shape})')
print(f'File full: {output_dir}/heart_disease_preprocessed.csv ({df_full_processed.shape})')
```
